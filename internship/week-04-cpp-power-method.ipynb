
# Power Method in C++ Called from Python for SVD Computation

## Objective
The objective of this task is to implement the Power Method in **C++** and call it
from **Python** to compute the **largest singular value** of a matrix.

The largest singular value is obtained by computing the dominant eigenvalue of
\( A^T A \) using the Power Method.


```python
import numpy as np
import power_method_cpp
```

```python
print(power_method_cpp)
```

```python
f = open('power_method.cpp', 'r')
file_contents = f.read()
print (file_contents)
```

```python
# Simple diagonal matrix
M = [[5.0, 0.0],
     [0.0, 1.0]]

dominant_eigenvalue = power_method_cpp.power_method(M, 1000, 1e-6)
print("Dominant eigenvalue:", dominant_eigenvalue)
```

```python
# Example matrix A
A = np.array([[1., 2.],
              [3., 4.],
              [5., 6.]])

# Compute A^T A
ATA = (A.T @ A).tolist()

# Largest eigenvalue of A^T A (C++ power method)
lambda_max = power_method_cpp.power_method(ATA, 1000, 1e-6)

# Largest singular value
sigma_max = np.sqrt(lambda_max)

print("Largest singular value (Power Method):", sigma_max)
```

```python

# NumPy SVD for comparison
sigma_numpy = np.linalg.svd(A, compute_uv=False)[0]
print("Largest singular value (NumPy):", sigma_numpy)
```


## Explanation

- The Power Method is implemented entirely in **C++**
- It is compiled as a Python extension using **pybind11**
- Python constructs the matrix \( A^T A \)
- The dominant eigenvalue of \( A^T A \) is computed in C++
- The largest singular value is obtained as:

$$
\sigma_{\max} = \sqrt{\lambda_{\max}(A^T A)}
$$

This satisfies the requirement that the Power Method be written in C++ and
called from Python.
